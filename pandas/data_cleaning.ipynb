```python
import pandas as pd
import numpy as np 
```

```python
df=pd.read_csv("employees_100.csv")
print(df)
```

```python
df.head()
```

```python
df["age"] = df["age"].fillna(df["age"].mean())
df
```

```python
df.isnull().sum()
```

```python
df["salary"]=df["salary"].str.replace(",","",regex=False)
df["salary"]=pd.to_numeric(df["salary"],errors="coerce") # bad values become NAN ,not a crash
df["salary"]= df["salary"].fillna(df["salary"].median())
```

```python
df.isnull().sum()
```

```python
df["department"].value_counts()
```

```python
df["department"]=df["department"].fillna(df["department"].mode()[0])
```

```python
df["city"]=df["city"].fillna(df["city"].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df=df.reset_index()
```

```python
df.drop_duplicates(subset=["employee_id"], keep="first")
```
